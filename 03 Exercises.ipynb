## Exercise 03.2

For the polynomial 
\begin{align}
f(x, y) &= (x + y)^{6} 
\\
&=  x^6 + 6x^{5}y + 15x^{4}y^{2} + 20x^{3}y^{3} + 15x^{2}y^{4} + 6xy^{5} + y^{6}
\end{align}
compute $f$ using: (i) the compact form $(x + y)^{6}$; and (ii) the expanded form for:

(a) $x = 10$ and $y = 10.1$

(b) $x = 10$ and $y = -10.1$

and compare the number of significant digits for which the answers are the same.
Store the answer for the compact version using the variable `f0`, and using the variable `f1` for the expanded version.

For case (b), compare the computed and analytical solutions and consider the relative error.
Which approach would you recommend for computing this expression?

#### (a) $x = 10$ and $y = 10.1$

```python
x = 10.0
y = 10.1

# compact version
f0 = (x + y)**6

# expanded version
f1 = x**6 + 6*(x**5)*(y) + 15*(x**4)*(y**2) + 20*(x**3)*(y**3) + 15*(x**2)*(y**4) + 6*(x)*(y**5) + y**6

# calculating absolute difference
df = abs(f0-f1)

# zeroing counter
n = 0

# converting scientific e.g. 2.3e-6 to string in standard form e.g. 0.00023
# then iterating over each character
for char in f"{df:.30f}":
    if char == '0' or char == '.':
        n += 1
    else:
        # first non zero character will be where f0 and f1 disagree
        # n will keep track of how many characters (sig fig) have been compared 
        print('The result is the same up to' ,n, 'significant figures')
        break
```

```python
## tests ##
import math
assert math.isclose(f0, 65944160.60120103, rel_tol=1e-10)
assert math.isclose(f1, 65944160.601201, rel_tol=1e-10)
```

#### (b) $x = 10$ and $y = -10.1$

```python
x = 10.0
y = -10.1

# compact version
f0 = (x + y)**6

# expanded version
f1 = x**6 + 6*(x**5)*(y) + 15*(x**4)*(y**2) + 20*(x**3)*(y**3) + 15*(x**2)*(y**4) + 6*(x)*(y**5) + y**6

# calculating absolute difference
df = abs(f0-f1)

# zeroing counter
n = 0

# converting scientific e.g. 2.3e-6 to string (using f"...") in standard form e.g. 0.00023 using fixed point notation
# and iterating over each character
for char in f"{df:.30f}":
    if char == '0' or char == '.':
        n += 1
    else:
        # first non zero character will be where f0 and f1 disagree
        # n will keep track of how many characters (sig fig) have been compared 
        print('The result is the same up to' ,n, 'significant figures')
        break

error = abs(f0 - f1)
print('The relative error is', error)

# better to use the compact format because for the expanded form it forces the computer to calculate large numbers
# which means it loses accuracy quickly (fewer d.p.)
```

```python
## tests ##
import math
assert math.isclose(f0, 1.0e-6, rel_tol=1e-10)
assert math.isclose(f1, 1.0e-6, rel_tol=1e-2)
```
